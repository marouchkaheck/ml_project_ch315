## 2. Split the data

Next, we split our data into a training set and a test set.

In order to prevent *any* information of the test set from leaking into our model, we split *before* starting to analyze or transform our data. For more details on why this matters, see [chapter 7.10.2 of Elements of Statistical Learning](https://web.stanford.edu/~hastie/ElemStatLearn//printings/ESLII_print10.pdf).

### 2.1. Split with stratification

[Stratification](https://en.wikipedia.org/wiki/Stratified_sampling) ensures that the class distributions (ratio of "good" to "bad" materials) are the same in the training and test set.

 $\color{DarkBlue}{\textsf{Short question}}$

**3.1)**  Why is this important? What could happen if we would not do this? 

Stratification is a way of splitting the data so that the proportion of each class stays the same in the training and the test set. Here, the classes are the low and the high performing materials. If 5% of all materials are high performing, about 5% of the training set and about 5% of the test set will be too.

This is important because a purely random split does not guarantee it. By chance, the test set could end up with only a few high performers, or the training set could. The model would then have little to learn from for the high uptake materials, or the test error would be computed on too few of them to mean anything. The results would also change a lot from one split to another.

For stratification to work, we need to define what makes a "good" or a "bad" material. We will use 15 mmol ${CO_2}$ / g as the threshold for the uptake, thus binarizing our continuous target variable. (You can choose it based on the histogram of the variables).

**5.1)** We chose 80% of the data for training and 20% for testing. A typical split is 70/30, as mentionned by the hints, but with 17379 materials we can afford to give more data to training and still keep 3476 materials for testing.

Choosing the split is a trade-off between bias and variance. With little training data, the model is worse than it could be, so the test error is pessimistically biased. With little test data, the estimate of the test error has a high variance, because it depends on which few materials ended up in the test set. For a small dataset, the training set would become too small with a 70/30 split, and for a huge dataset, a 30% test set would be larger than needed, since a few percent is already thousands of points. In the extreme cases, a single test point would give an error that says nothing about the model in general, and having more test points than training points would leave the model without enough data to learn from, so its performance would drop.

**5.2)** We split the data to be able to judge the model on materials it has not seen. If we evaluated it on the training set, the error would look better than it really is, because the model was fitted on exactly these points and can simply memorise them. The test set is kept aside and never used for fitting, so the error computed on it tells us how well the model works on new materials, and shows whether it overfits.

**5.3)** A hyperparameter is a setting of the model that is not learned from the data but has to be chosen before the training. To choose it, we train the model with different values and compare how well each one performs. This comparison must not be done on the test set. If it were, the test set would influence which model we keep, so the test error would no longer be independent of the model and would be too optimistic. The hyperparameters are instead tuned on a separate validation set, or by cross-validation on the training set, and the test set is used only once, at the end, to evaluate the final model.
In summary, no, the test set must not be used to tune the hyperparameters of our model.

## 4. Baselines

For machine learning, it is important to have some *baselines* to which one then compares the results of a model. Think of a classification model for some rare disease where we only have 1% postives. A classification model that only predictes the negatives *all the time* will still have a amazingly high accuracy. To be able to understand if our model is really better than such a simple prediction we need to make the simple prediction first. This is what we call a baseline.

A baseline could be a really simple model, a basic heuristic or the current state of the art.
this. We will use a heuristic.

For this we use sklearn `Dummy` objects that simply calculate the mean, the median or the most frequent case of the training set, when you run the `fit()` method on them (which takes the features matrix $\mathbf{X}$ and the labels $\mathbf{y}$ as arguments.
This is, the prediction of a `DummyRegressor` with `mean` strategy will always be the mean, independent of the input (it will not look at the feature matrix!). 

Instead of using those `sklearn` objects you could also just manually compute the the mean or median of the dataset. But we will use those objects as we can learn in this way how to use estimators in `sklearn` and it is also allows you to test your full pipeline with different (baseline) models. 
What does this mean? In practice this means that you can use all the regression and classification models shown in the figure below in the same way, they will all have a `fit()` method that accepts `X` and `y` and a predict method that accepts `X` and returns the predictions.

You will now write functions `get_regression_metrics(model, X, y_true)` that compute the metrics and return this dictionary for a given model. The `predict` method takes the feature matrix $\mathbf{X}$ as input.

In them, we calculate 

$\mathrm {MAE} =\frac{\sum _{i=1}^{n}\left|Y_{i}-\hat{y}_{i}\right|}{n}.$

and 

$\mathrm {MSE} = {\frac {1}{n}}\sum _{i=1}^{n}(Y_{i}-{\hat {Y_{i}}})^{2}.$ 

where $\hat{y}$ are the predictions and, $Y_{i}$ the true values.

as well as the maximum error.

<details>
<summary> <font color='green'>Click here for hints</font></summary>

    <li> to perform a prediction using a estimator object, you can call <code> classifier.predict(X) </code> </li>

````python
def get_regression_metrics(model, X, y_true):
    """
    Get a dicionary with regression metrics:

    model: sklearn model with predict method
    X: feature matrix
    y_true: ground truth labels
    """
    y_predicted = model.predict(X)

    mae = mean_absolute_error(y_true, y_predicted)
    mse = mean_squared_error(y_true, y_predicted)
    maximum_error = max_error(y_true, y_predicted)

    metrics_dict = {
        'mae': mae,
        'mse': mse,
        'max_error': maximum_error
    }

    return metrics_dict
````

**9.1)** Three metrics are used to compare the predictions $\hat{y}_i$ with the true uptakes $Y_i$ of $n$ materials:

- The mean absolute error (MAE), $\frac{1}{n}\sum_{i=1}^{n}|Y_i - \hat{y}_i|$, is the average size of the error. It is given in mmol/g.
- The mean squared error (MSE), $\frac{1}{n}\sum_{i=1}^{n}(Y_i - \hat{y}_i)^2$, is the average of the squared errors, which gives more weight to large errors. It is given in (mmol/g)².
- The maximum error, $\max_i |Y_i - \hat{y}_i|$, is the largest error made on a single material. It is given in mmol/g.

The function `get_regression_metrics(model, X, y_true)` predicts the uptakes of the materials with the model and returns these three metrics in a dictionary. It is applied to each dummy model on the training and on the test set.

| | Mean (train) | Median (train) | Mean (test) | Median (test) |
|---|---|---|---|---|
| MAE | 6.33 | 6.26 | 6.24 | 6.16 |
| MSE | 54.36 | 56.36 | 52.38 | 54.20 |
| Max error | 22.69 | 24.11 | 20.95 | 22.36 |

The two dummy models predict a constant, so the errors are expected to be large, of the order of the spread of the uptakes, and similar on the training and the test set. For the mean model, the training MSE should equal the variance of the training uptakes, and the maximum error should be close to the distance between the constant and the most extreme uptake. On the test set, the MSE should be close to the test variance plus the squared difference between the test and training means. The mean minimises the MSE and the median minimises the MAE, so the median model should have the lowest MAE and the highest MSE of the two.

**9.2)** The values are not surprising. They match the expectations of the previous question: the errors are similar on the training and the test set, and the median model has the lowest MAE while the mean model has the lowest MSE on both sets. The small differences between the two sets (an MAE of 6.24 on the test set against 6.33 mmol/g on the training set for the mean model) come from the sampling of the data.

The training MSE of the mean model (54.36 (mmol/g)²) equals the variance of the training uptakes, so its square root, about 7.4 mmol/g, is their standard deviation. The typical error of a constant prediction is therefore of the size of the spread of the data itself, as expected when the features are not used. This is large compared with the 15 mmol/g threshold separating low and high performing materials, so a constant prediction cannot distinguish good materials from poor ones. The maximum errors, between 21 and 24 mmol/g, correspond to the materials with the largest uptakes of the dataset, which are far above the constant prediction.

**9.3)** A metric only means something when it is compared with a reference. An MAE of 6 mmol/g is neither good nor bad on its own, but it becomes informative once it is known that a model predicting the mean already reaches an MAE of 6.3 mmol/g. Baseline models give this reference: they show the error obtained without using any information on the materials. A model is useful only if its errors are clearly lower than those of the baselines.

In practice, the metrics of a model should always be reported next to those of the baselines, since the improvement over the baselines quantifies what the model has actually learned. If a model does not beat them, it has learned nothing from the descriptors, or there is a mistake in the workflow. This also protects against misleading results, such as a high accuracy obtained on an imbalanced dataset by always predicting the majority class.

$\color{DarkBlue}{\textsf{Short Question}}$
- Why might the scaling of the features be relevant for a machine learning model? 

**12)** Feature scaling is important because the descriptors used in the dataset have different units and numerical ranges, such as density, pore dimensions and surface area. In Kernel Ridge Regression, particularly with an RBF kernel, similarities between materials depend on the distances between their feature vectors. Without scaling, features with larger numerical values may dominate these distances, even if they are not necessarily more relevant for predicting gas uptake.

To address this, we use `StandardScaler`, which transforms each feature by subtracting its mean and dividing by its standard deviation, resulting in features with a mean of zero and a standard deviation of one. This allows features to contribute on more comparable scales and can improve the stability and predictive performance of the model.

The most common approach to hyperparameter optimization is to define a grid of all relevant parameters and to search over the grid for the best model performance.

## 8. Feature Engineering 

Finally, we would like to remove features with low variance. This can be done by setting a variance threshold.

$\color{DarkBlue}{\textsf{Short Question}}$
    
- What is the reasoning behind doing this? 
- When might it go wrong and why?

**15.1)** Features with very low variance have nearly constant values across the dataset and therefore provide little information for distinguishing between materials. Removing these features reduces the dimensionality of the dataset, simplifies the model and can decrease computational cost. It may also reduce unnecessary noise and improve model performance, particularly when dealing with a large number of descriptors.

**15.2)** Removing low-variance features can be problematic because low variance does not necessarily imply low predictive importance. For example, a rare chemical feature might be present in only a few MOFs but have a significant influence on their gas adsorption capacity. Removing this descriptor could therefore eliminate valuable information and reduce prediction accuracy.
Furthermore, since variance depends on the numerical scale of the features, applying a threshold without considering feature scaling could remove relevant descriptors simply because their values are small. Therefore, variance-based feature selection should be evaluated carefully to ensure that important information is not lost.

## 10. Influence of Regularization

 $\color{DarkBlue}{\textsf{Short Exercise}}$

**19.1)** To investigate the effect of the regularization parameter $\alpha$ in Kernel Ridge Regression, we will compare two models with very different values of $\alpha$: one very small and one very large.
We will use the scaled KRR pipeline, keeping $\alpha$ fixed for each model while optimizing the RBF kernel parameter $\gamma$ using cross-validation. We will then compare the training and test errors to understand how the regularization strength affects model performance.


In Ridge Regression, the model minimizes an objective function composed of two terms:
$$
\min_{\mathbf{w}}\left[
\underbrace{\sum_{i=1}^{n}(y_i-\hat{y}_i)^2}_{\text{Prediction error}}
+
\underbrace{\alpha|\mathbf{w}|_2^2}_{\text{Regularization penalty}}
\right]
$$
The first term measures how well the model fits the training data, while the second penalizes large model coefficients. The parameter $\alpha$ controls the importance of this penalty.

•	Very small $\alpha$: The regularization penalty is weak, allowing the model to fit the training data more closely, but increasing the risk of overfitting.

•	Very large $\alpha$: The penalty becomes stronger, restricting the model's flexibility and potentially causing underfitting.
For Kernel Ridge Regression, the corresponding dual coefficients are calculated using:
$$
\boldsymbol{\beta}=(K+\alpha I)^{-1}\mathbf{y}
$$
where $K$ is the kernel matrix, $I$ is the identity matrix and $\mathbf{y}$ contains the target values.

As discussed when studying the loss landscapes of LASSO, adding a regularization penalty changes the shape of the objective function and therefore the position of its minimum. Increasing $\alpha$ gives more weight to this penalty, encouraging the model to choose smaller coefficients, even if this means accepting a larger training error.

We compare four values of the KRR regularization parameter $\alpha$, ranging from very weak to very strong regularization. For each value, we optimize the RBF kernel parameter $\gamma$ using 3-fold cross-validation to obtain the best possible performance within our search range, rather than having a poorly chosen $\gamma$ influence the results. We then calculate the training and test errors to investigate how regularization affects model performance.

````python
from sklearn.base import clone
from sklearn.model_selection import GridSearchCV

alpha_values = [0.0001, 0.01, 1, 100]
gamma_values = [0.001, 0.01, 0.1]

results_alpha = {}

for alpha in alpha_values:
    pipe_alpha = clone(pipe_w_scaling)
    pipe_alpha.set_params(krr__alpha=alpha)

    search_alpha = GridSearchCV(
        pipe_alpha,
        param_grid={"krr__gamma": gamma_values},
        cv=3,
        scoring="neg_mean_absolute_error",
        n_jobs=1
    )

    search_alpha.fit(
        df_train_stratified[FEATURES],
        df_train_stratified[TARGET]
    )

    best_model = search_alpha.best_estimator_

    results_alpha[alpha] = {
        "gamma": search_alpha.best_params_["krr__gamma"],
        "train": get_regression_metrics(
            best_model,
            df_train_stratified[FEATURES],
            df_train_stratified[TARGET]
        ),
        "test": get_regression_metrics(
            best_model,
            df_test_stratified[FEATURES],
            df_test_stratified[TARGET]
        )
    }

    print(f"\nAlpha = {alpha}", flush=True)
    print("Best gamma:", results_alpha[alpha]["gamma"], flush=True)
    print("Training:", results_alpha[alpha]["train"], flush=True)
    print("Test:", results_alpha[alpha]["test"], flush=True)
````

The results show that the regularization parameter $\alpha$ has a significant effect on KRR performance.

The best test performance was obtained with $\alpha = 0.0001$, giving a MAE of 0.724 and a MSE of 1.284. With $\alpha = 0.01$, the training error was slightly lower, but the test error increased, suggesting a greater tendency to overfit.

As $\alpha$ increases, both training and test errors generally become larger. This is particularly clear for $\alpha = 100$, where the test MAE reaches 2.476 and the MSE reaches 17.393. This indicates that excessive regularization restricts the model too much, leading to underfitting.

Overall, the results illustrate the importance of choosing an appropriate regularization strength: a small $\alpha$ allows the model to fit the data more closely, while an excessively large $\alpha$ reduces its ability to capture the relationship between MOF descriptors and CO₂ uptake.

- **21)** Complete the function `_calculate_permutation_scores` (which we took from the `sklearn` package) and which is needed to calculate the permutation feature importance using the `permutation_importance` function. 

````python
def _calculate_permutation_scores(estimator, X, y, col_idx, random_state,
                                  n_repeats, scorer):
    """Calculate score when `col_idx` is permuted. Based on the sklearn implementation

    estimator: sklearn estimator object
    X: pd.Dataframe or np.array
    y: pd.Dataframe or np.array
    col_idx: int
    random_state: int
    n_repeats: int
    scorer: function that takes model, X and y_true as arguments
    """
    random_state = check_random_state(random_state)

    X_permuted = X.copy()
    scores = np.zeros(n_repeats)
    shuffling_idx = np.arange(X.shape[0])

    for n_round in range(n_repeats):

        # Shuffle the row indices
        random_state.shuffle(shuffling_idx)

        # Deal with dataframes
        if hasattr(X_permuted, "iloc"):
            col = X_permuted.iloc[shuffling_idx, col_idx]
            col.index = X_permuted.index
            X_permuted.iloc[:, col_idx] = col

        # Deal with numpy arrays
        else:
            X_permuted[:, col_idx] = X_permuted[shuffling_idx, col_idx]

        # Get the scores
        feature_score = scorer(estimator, X_permuted, y)
        scores[n_round] = feature_score

    return scores
````

Nothing to change in the function below, it just call the `_calculate_permutation_scores` function you just completed. 

````python
def permutation_importance(
    estimator,
    X,
    y,
    scoring="neg_mean_absolute_error",
    n_repeats=5,
    n_jobs=2,
    random_state=None,
):
    """Permutation importance for feature evaluation
    estimator : object
        An estimator that has already been :term:`fitted` and is compatible
        with :term:`scorer`.
    X : ndarray or DataFrame, shape (n_samples, n_features)
        Data on which permutation importance will be computed.
    y : array-like or None, shape (n_samples, ) or (n_samples, n_classes)
        Targets for supervised or `None` for unsupervised.
    scoring : string, callable or None, default=None
        Scorer to use. It can be a single
        string (see :ref:`scoring_parameter`) or a callable (see
        :ref:`scoring`). If None, the estimator's default scorer is used.
    n_repeats : int, default=5
        Number of times to permute a feature.
    n_jobs : int or None, default=2
        The number of jobs to use for the computation.
        `None` means 1 unless in a :obj:`joblib.parallel_backend` context.
        `-1` means using all processors. See :term:`Glossary <n_jobs>`
        for more details.
    random_state : int, RandomState instance, or None, default=None
        Pseudo-random number generator to control the permutations of each
        feature. See :term:`random_state`.
    """
    # Deal with dataframes
    if not hasattr(X, "iloc"):
        X = check_array(X, force_all_finite="allow-nan", dtype=None)

    # Precompute random seed from the random state to be used
    # to get a fresh independent RandomState instance for each
    # parallel call to _calculate_permutation_scores, irrespective of
    # the fact that variables are shared or not depending on the active
    # joblib backend (sequential, thread-based or process-based).
    random_state = check_random_state(random_state)
    random_seed = random_state.randint(np.iinfo(np.int32).max + 1)

    # Determine scorer from user options.
    scorer = check_scoring(estimator, scoring=scoring)
    # get the performance score on the unpermuted data
    baseline_score = scorer(estimator, X, y)

    # run the permuted evaluations in parallel for each column
    scores = Parallel(n_jobs=n_jobs)(
        delayed(_calculate_permutation_scores)(
            estimator, X, y, col_idx, random_seed, n_repeats, scorer
        )
        for col_idx in range(X.shape[1])
    )

    # get difference two
    importances = baseline_score - np.array(scores)

    # return the results (dictionary)
    return Bunch(
        importances_mean=np.mean(importances, axis=1),
        importances_std=np.std(importances, axis=1),
        importances=importances,
    )
````
